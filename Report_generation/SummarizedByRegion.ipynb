```python
from pathlib import Path
import pandas as pd
import logging

PROJECT_DIR = Path.cwd().parent
input_file = PROJECT_DIR / "Report_generation" / "data_with_revenue.csv"

try:
    df = pd.read_csv(input_file)

    required_columns = [
        "Region", "Quantity", "Unit Price", "Order ID"
    ]

    missing = [col for col in required_columns if col not in df.columns]

    if missing:
        raise ValueError(f"Missing columns: {missing}")

    df["Revenue"] = df["Quantity"] * df["Unit Price"]

    logging.info("Revenue CSV loaded successfully for region analysis")
    print("Data loaded successfully")
    display(df.head())

except FileNotFoundError:
    logging.exception("Revenue CSV file not found")
    raise

except Exception:
    logging.exception("Failed to load data for region analysis")
    raise

logging.basicConfig(
    filename=str(PROJECT_DIR / "sales_analysis.log"),
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("Region analysis started")
```

```python
try:
    region_summary = df.groupby("Region").agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Total_Orders=("Order ID", "nunique")
    ).reset_index()

    region_summary = region_summary.sort_values(
        by="Total_Revenue",
        ascending=False
    )

    display(region_summary)

    logging.info("Region summary generated successfully")

except Exception:
    logging.exception("Region summary generation failed")
    raise
```

```python
# Summarize sales by region
region_summary = df.groupby("Region").agg(
    Total_Revenue=("Revenue", "sum"),
    Total_Quantity=("Quantity", "sum"),
    Total_Orders=("Order ID", "nunique")
).reset_index()
region_summary.head()
```

```python
# Sort by revenue from highest to lowest
region_summary = region_summary.sort_values(
    by="Total_Revenue", ascending=False
)
region_summary.head(3)
```

```python
# Display the summary
print(region_summary.to_string(index=False))
```
