```python
from pathlib import Path
import pandas as pd
import logging

PROJECT_DIR = Path.cwd().parent
input_file = PROJECT_DIR / "Report_generation" / "data_with_revenue.csv"

try:
    df = pd.read_csv(input_file)

    logging.info("Revenue CSV loaded for ranking reports")
    print("CSV loaded successfully")
    display(df.head())

except FileNotFoundError:
    logging.exception("Revenue CSV not found")
    raise

except Exception:
    logging.exception("Failed to load revenue data")
    raise
logging.basicConfig(
    filename=str(PROJECT_DIR / "sales_analysis.log"),
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("Product and salesperson analysis started")
```

```python
Top_products = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
)


print("Top 3 Products by Revenue:")
print(Top_products)
```

```python
try:
    top_products = (
        df.groupby("Product")["Revenue"]
        .sum()
        .sort_values(ascending=False)
        .head(5)
    )

    top_salespersons = (
        df.groupby("Salesperson")["Revenue"]
        .sum()
        .sort_values(ascending=False)
        .head(5)
    )

    print("Top Products:")
    display(top_products)

    print("Top Salespersons:")
    display(top_salespersons)

    logging.info("Product and salesperson rankings generated")

except Exception:
    logging.exception("Ranking report generation failed")
    raise
```
