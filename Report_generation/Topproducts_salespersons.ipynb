```python
import pandas as pd
```

```python
df = pd.read_csv("C:\\Users\\Admin\\Desktop\\Final Project\\Report_generation\\data_with_revenue.csv")
df.info()
```

```python
Top_products = (
    df.groupby("Product")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(3)
)


print("Top 3 Products by Revenue:")
print(Top_products)
```

```python
top_salespersons = (
    df.groupby("Salesperson")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print("\nTop 5 Salespersons by Revenue:")
print(top_salespersons)
```
