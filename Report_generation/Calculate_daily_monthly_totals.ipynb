```python
from pathlib import Path
import pandas as pd
import logging

PROJECT_DIR = Path.cwd().parent
input_file = PROJECT_DIR / "Report_generation" / "data_with_revenue.csv"

logging.basicConfig(
    filename=str(PROJECT_DIR / "sales_analysis.log"),
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("Daily and monthly sales analysis started")
```

```python


try:
    df = pd.read_csv(input_file)
    logging.info("Sales data loaded successfully")
    print("Data loaded successfully")

except FileNotFoundError:
    logging.exception("Input CSV file not found")
    raise

except Exception:
    logging.exception("Failed to load sales data")
    raise

df["Date"] = pd.to_datetime(df["Date"], dayfirst=True, errors="coerce")
```

```python
try:
    required_columns = ["Date", "Quantity", "Unit Price", "Order ID"]
    missing = [col for col in required_columns if col not in df.columns]

    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    df["Date"] = pd.to_datetime(
        df["Date"],
        dayfirst=True,
        errors="coerce"
    )

    df["Revenue"] = df["Quantity"] * df["Unit Price"]
    df = df.dropna(subset=["Date"])

    logging.info("Date conversion and revenue calculation completed")

except Exception:
    logging.exception("Date or revenue calculation failed")
    raise
```

```python
df = df.dropna(subset=["Date"])
```

```python
# Calculate daily sales totals
try:
    daily_totals = df.groupby(df["Date"].dt.date).agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Total_Orders=("Order ID", "nunique")
    ).reset_index()

    daily_totals = daily_totals.rename(columns={"Date": "Day"})

    print("DAILY SALES TOTALS")
    print(daily_totals.to_string(index=False))

    daily_file = PROJECT_DIR / "Report_generation" / "daily_sales_totals.csv"
    daily_totals.to_csv(daily_file, index=False)

    logging.info("Daily sales totals saved successfully")

except Exception:
    logging.exception("Daily totals calculation failed")
    raise
```

```python
daily_totals.to_csv("daily_sales_totals.csv", index=False)
```

```python
#monthly sales totals
try:
    df["Month"] = df["Date"].dt.to_period("M")

    monthly_totals = df.groupby("Month").agg(
        Total_Revenue=("Revenue", "sum"),
        Total_Quantity=("Quantity", "sum"),
        Total_Orders=("Order ID", "nunique")
    ).reset_index()

    print("\nMONTHLY SALES TOTALS")
    print(monthly_totals.to_string(index=False))

    monthly_file = PROJECT_DIR / "Report_generation" / "monthly_sales_totals.csv"
    monthly_totals.to_csv(monthly_file, index=False)

    logging.info("Monthly sales totals saved successfully")

except Exception:
    logging.exception("Monthly totals calculation failed")
    raise
```

```python
monthly_totals.to_csv("monthly_sales_totals.csv", index=False)
```
