```python
import pandas as pd 
# df = pd.read_csv("C:\\Users\\Admin\\Desktop\\Final Project\\Data_sets\\messy_sales_data.csv")
# df.head()
import logging
from pathlib import Path

try:
    file_path = PROJECT_DIR / "Data_sets" / "messy_sales_data.csv"
    df = pd.read_csv(file_path)

    logging.info("Raw sales CSV loaded successfully")
    print(df.head())

except FileNotFoundError:
    logging.exception("Raw sales CSV file not found")
    raise

except Exception:
    logging.exception("Error while loading sales data")
    raise

logging.basicConfig(
    filename=str(PROJECT_DIR / "sales_analysis.log"),
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    force=True
)

logging.info("Data cleaning started")


```

```python
print(pd.isnull(df))
print("\nMissing values per column:")
print(df.isnull().sum())
```

```python
df.duplicated().sum()   
```

```python
before = len(df)

df.drop_duplicates(inplace=True)

removed = before - len(df)

logging.info("Duplicate rows removed: %s", removed)

print("Duplicate rows removed:", removed)
print("Remaining duplicates:", df.duplicated().sum())
from pathlib import Path

output_file = Path.cwd() / "cleaned_data.csv"
df.to_csv(output_file, index=False)
print("Cleaned data saved:", output_file.exists())
print("Cleaned data location:", output_file)

print("Cleaned sales dataset saved!")
print("Total rows after removing duplicates:", len(df))
print("\nMissing values per column after cleaning:")
print(df.isnull().sum())
print("\nFirst 10 rows after cleaning:")
print(df.head(10))
```

```python
df['Salesperson'].isna().sum()
df.dropna(subset=['Salesperson'], inplace=True) 
```

```python
df["Date"] = df["Date"].ffill(inplace=True)
```

```python
df['Date'].isna().sum()
```

```python
df["Region"] = df["Region"].bfill(inplace=True)
```

```python
df['Region'].isna().sum()
```

```python
df["Quantity"]=df["Quantity"].fillna(df["Quantity"].mean(), inplace=True)
```
